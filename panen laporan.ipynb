```python
def buat_laporan(nama, jumlah, harga):
    total = jumlah * harga

    print("\n===== LAPORAN HASIL PANEN =====")
    print("Jenis hasil panen :", nama)
    print("Jumlah panen      :", jumlah, "kg")
    print("Harga per kg      : Rp", harga)
    print("Total hasil panen : Rp", total)
```
